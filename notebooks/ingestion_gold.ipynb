```python
import pandas as pd

from pathlib import Path
```

```python
silver_path = Path("../data/silver/bank_marketing_silver.csv")

df = pd.read_csv(silver_path, sep=";")

print(f"Dados carregados: {df.shape[0]} linhas e {df.shape[1]} colunas")
df.head()
```

```python
if "duration" in df.columns:
    df = df.drop(columns=["duration"])
    print("Coluna 'duration' removida.")
else:
    print("A coluna 'duration' não existe no DataFrame.")

if "default" in df.columns:
    df = df.drop(columns=["default"])
    print("Coluna 'default' removida.")
else:
    print("A coluna 'default' não existe no DataFrame.")
```

```python
df["never_contacted"] = (df["pdays"] == 999).astype(int)
```

```python
df["campaign_bucket"] = pd.cut(
    df["campaign"],
    bins=[0, 1, 3, float("inf")],
    labels=["first_contact", "few_contacts", "many_contacts"]
)
```

```python
df["previous_success"] = (
    df["poutcome"] == "success"
).astype(int)
```
