```python
import pandas as pd
import json
import kagglehub

from pathlib import Path
from datetime import datetime
```

```python
path = kagglehub.dataset_download("henriqueyamahata/bank-marketing")
```

```python
files = [p for p in Path(path).iterdir() if p.is_file()]

csv_files = [p for p in files if p.suffix.lower() == ".csv"]

if not csv_files:
    raise FileNotFoundError(f"Nenhum arquivo CSV encontrado em: {path}")

preferred = [p for p in csv_files if "additional-full" in p.name.lower()]
csv_path = preferred[0] if preferred else csv_files[0]
print(f"Arquivo selecionado: {csv_path.name}")

df = pd.read_csv(csv_path, sep=";")

if df.empty:
    raise ValueError("Dataset vazio.")

expected_columns = [
    'age',
    'job',
    'marital',
    'education', 
    'default', 
    'housing', 
    'loan',
    'contact', 
    'month', 
    'day_of_week', 
    'duration', 
    'campaign', 
    'pdays',
    'previous', 
    'poutcome', 
    'emp.var.rate', 
    'cons.price.idx',
    'cons.conf.idx', 
    'euribor3m', 
    'nr.employed', 
    'y'
]

assert list(df.columns) == expected_columns

df.head()
```

```python
df.columns
```

```python
output_dir = Path("../data/bronze")
output_dir.mkdir(parents=True, exist_ok=True)

output_path = output_dir / "bank_marketing_bronze.csv"
df.to_csv(output_path, index=False, sep=";")

print(f"Arquivo salvo em: {output_path}")
```

```python
metadata = {
    "dataset_name": "Bank Marketing",
    "source": "Kaggle",
    "dataset_id": "henriqueyamahata/bank-marketing",
    "download_method": "kagglehub",
    "original_file": "bank-additional-full.csv",
    "target": "y",
    "rows": len(df),
    "columns": len(df.columns),
    "download_date": datetime.now().strftime("%Y-%m-%d"),
    "bronze_version": "1.0"
}

output_dir = Path("../data/kaggle")
output_dir.mkdir(parents=True, exist_ok=True)

output_path = output_dir / "metadata.json"

with open(output_path, "w") as f:
    json.dump(metadata, f, indent=4)
```
