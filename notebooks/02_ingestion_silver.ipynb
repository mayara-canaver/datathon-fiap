```python
import pandas as pd
import re

from pathlib import Path
```

```python
bronze_path = Path("../data/bronze/bank_marketing_bronze.csv")

df = pd.read_csv(bronze_path, sep=";")

print(f"Dados carregados: {df.shape[0]} linhas e {df.shape[1]} colunas")
df.head()
```

```python
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]

print(f"Linhas antes: {before}")
print(f"Linhas após remover duplicados: {after}")
print(f"Duplicatas removidas: {before - after}")
```

```python
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].astype("string")

# Ajuste de tipagem para colunas numéricas comuns
for col in ["age", "balance", "day", "duration", "campaign", "pdays", "previous"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Ajuste de tipagem para colunas booleanas comuns
for col in ["default", "housing", "loan", "y"]:
    if col in df.columns:
        df[col] = df[col].astype("string")

df.dtypes
```

```python
df.default.value_counts()
```

```python
for col in ["job", "education"]:
    if col in df.columns:
        df[col] = df[col].astype("string")
        df[col] = df[col].str.replace(r"[\.-]", " ", regex=True)
        df[col] = df[col].str.strip()

for col in df.columns:
    if df[col].dtype == "string":
        df[col] = df[col].str.strip()
```

```python
df.isna().sum()
```

```python
mask = (df["previous"] == 0) & (df["poutcome"].astype(str).str.lower() == "success")

if mask.any():
    display(df.loc[mask].head())
```

```python
output_dir = Path("../data/silver")
output_dir.mkdir(parents=True, exist_ok=True)

output_path = output_dir / "bank_marketing_silver.csv"
df.to_csv(output_path, index=False, sep=";")

print(f"Arquivo salvo em: {output_path}")
```
